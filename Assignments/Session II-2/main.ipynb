```python
import pandas as pd
# from sklearn.model_selection import train_test_split
# from sklearn.naive_bayes import GaussianNB
```

```python
dataset = pd.read_csv('./Naive-Bayes-Classification-Data.csv')
```

```python
# Calculate probabilities for Naive Bayes manually
import numpy as np

class_0 = dataset[dataset['diabetes'] == 0]
class_1 = dataset[dataset['diabetes'] == 1]

p_class_0 = len(class_0) / len(dataset)
p_class_1 = len(class_1) / len(dataset)

mean_var_0 = {col: (class_0[col].mean(), class_0[col].var()) for col in ['glucose', 'bloodpressure']}
mean_var_1 = {col: (class_1[col].mean(), class_1[col].var()) for col in ['glucose', 'bloodpressure']}
```

```python
def gaussian_pdf(x, mean, var):
    return (1 / np.sqrt(2 * np.pi * var)) * np.exp(-((x - mean) ** 2) / (2 * var))
```

```python
def predict(sample):
    likelihood_0 = p_class_0
    for factor in ['glucose', 'bloodpressure']:
        mean, var = mean_var_0[factor]
        likelihood_0 *= gaussian_pdf(sample[factor], mean, var)

    likelihood_1 = p_class_1
    for factor in ['glucose', 'bloodpressure']:
        mean, var = mean_var_1[factor]
        likelihood_1 *= gaussian_pdf(sample[factor], mean, var)

    total = likelihood_0 + likelihood_1
    return {'class_0': likelihood_0 / total, 'class_1': likelihood_1 / total}
```

```python
glucose, blood_pressure = map(int,input().split())
sample1 = {'glucose': glucose, 'bloodpressure': blood_pressure}
probabilities = predict(sample1)
print(sample1)
print(f"P(Diabetes) = {probabilities['class_1']}")
print(f"P(Diabetes*) = {probabilities['class_0']}")
print(f"Diabetes? {'Yes' if probabilities['class_1']>probabilities['class_0'] else 'No'}")
```
