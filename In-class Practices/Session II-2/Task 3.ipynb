```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

**Defining** **Dataset**

```python
# Assigning features "Continuous varaibles" and label variables
height=[6,5.92,5.58,5.92,5,5.5,5.42,5.75]   # in feet (ft)
weight=[180,190,170,165,100,150,130,150]    # in bounds (lbs)
foot = [12,11,12,10,6,8,7,9]                # in inch (in)
sex=['male','male','male','male','female','female','female','female']  # label male or female
```

**Gaussian** **Distribution**

```python
# calculate average and variance for male height
htm = np.array(height[:4])     # Height male
htmmean = np.average(htm)      # Height male mean
htmsegma = np.std(htm)         # std: standrad deviation   # var: Variance    # Height male standard deviation
#print('Mean: ',htmmean)       # Just to check
#print('Segma: ',htmsegma)
```

```python
# calculate average and variance for female height
htf = np.array(height[4:])     # Height female
htfmean = np.average(htf)      # Height female mean
htfsegma = np.std(htf)         # std: standrad deviation   # var: Variance  # height female standard deviation
#print('Mean: ',htfmean)       # Just to check
#print('Segma: ',htfsegma)
```

```python
# calculate average and variance for male weight
wtm = np.array(weight[:4])    # Weight male
wtmmean = np.average(wtm)     # weight male mean
wtmsegma = np.std(wtm)        # std: standrad deviation   # var: Variance  # weight male standard deviation
#print('Mean: ',wtmmean)      # Just to check
#print('Segma: ',wtmsegma)
```

```python
# calculate average and variance for female weight
wtf = np.array(weight[4:])    #Weight female
wtfmean = np.average(wtf)     #Weight female mean
wtfsegma = np.std(wtf)        # std: standrad deviation   # var: Variance #Weight female standard deviation
#print('Mean: ',wtfmean)      # Just to check
#print('Segma: ',wtfsegma)
```

```python
# calculate average and variance for male foot
ftm = np.array(foot[:4])       #Foot male
ftmmean = np.average(ftm)      #Foot male mean
ftmsegma = np.std(ftm)         # std: standrad deviation   # var: Variance   #Foot male standard deviation
#print('Mean: ',ftmmean)       # Just to check
#print('Segma: ',ftmsegma)
```

```python
# calculate average and variance for female foot
ftf = np.array(foot[4:])      #foot Female
ftfmean = np.average(ftf)     #foot Female mean
ftfsegma = np.std(ftf)        # std: standrad deviation   # var: Variance  #foot Female standard deviation
#print('Mean: ',ftfmean)      # Just to check
#print('Segma: ',ftfsegma)
```

```python
# calculate probability Density function PDF to be male
phm = 1/(np.sqrt(2*np.pi)*htmsegma)*np.exp(-(6-htmmean)**2/(2*htmsegma**2))   #pdf for height male
pwm = 1/(np.sqrt(2*np.pi)*wtmsegma)*np.exp(-(130-wtmmean)**2/(2*wtmsegma**2)) #pdf for weight male
pfm = 1/(np.sqrt(2*np.pi)*ftmsegma)*np.exp(-(8-ftmmean)**2/(2*ftmsegma**2))   #pdf for foot male
probm = 0.5*phm*pwm*pfm      # probability to be male : note: this is the numerator as the denominator is common
print(phm,'\t',pwm,'\t',pfm)
print("Probability to be male: ",probm)
```

```python
# calculate probability Density function PDF to be female
phf = 1/(np.sqrt(2*np.pi)*htfsegma)*np.exp(-(6-htfmean)**2/(2*htfsegma**2))    #pdf for height female
pwf = 1/(np.sqrt(2*np.pi)*wtfsegma)*np.exp(-(130-wtfmean)**2/(2*wtfsegma**2))  #pdf for weight female
pff = 1/(np.sqrt(2*np.pi)*ftfsegma)*np.exp(-(8-ftfmean)**2/(2*ftfsegma**2))    #pdf for foot female
probf = 0.5*phf*pwf*pff     # probability to be female : note: this is the numerator as the denominator is common
print(phf,'\t',pwf,'\t',pff)
print("Probability to be female: ",probf)
```
