```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = 2*np.random.rand(100,1)
y = 4 + 3*x + np.random.rand(100,1)
```

```python
plt.plot(x,y,'b.')
plt.xlabel("$x_1$",fontsize=18)
plt.ylabel("$y$",rotation=0,fontsize=18)
plt.axis([0,2,0,15])
plt.show()
```

```python
from sklearn.model_selection import train_test_split,KFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
linear_regression = LinearRegression()
linear_regression.fit(x,y)
```

```python
plt.plot(x, linear_regression.predict(x), 'r-', label='Linear Regression Line')
plt.plot(x,y,'b.')
plt.xlabel("$x_1$",fontsize=18)
plt.ylabel("$y$",rotation=0,fontsize=18)
plt.axis([0,2,0,15])
plt.show()
```

```python
data = pd.read_csv('./diabetes.csv')
x = data.iloc[:,:-1].to_numpy()
y = data.iloc[:,-1].to_numpy()

print(type(x),type(y))
print(x.shape,y.shape)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
print(x_train[0],'\n',y_train[0])
print(x_train.shape,x_test.shape)
```

```python
linear_regression = LinearRegression()
model = linear_regression.fit(x_train,y_train)
y_pred = model.predict(x_test)
err = mean_squared_error(y_test,y_pred)
print(err)

y_pred = model.predict(x_train)

X_df = pd.DataFrame(x_train, columns=[f'x{i}' for i in range(x_train.shape[1])])
for col in X_df.columns:
    plt.figure()
    plt.scatter(X_df[col], y_pred, alpha=0.5)
    # plt.plot(X_df[col],y_pred,'r--')
    plt.xlabel(col)
    plt.ylabel('y')
    plt.title(f'{col} vs y')
    plt.show()
```
