```python
x = np.array([i for i in range(0,6)])
y = np.array([0,8,12,20,26,38])

plt.plot(x,y)
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Sport Data')
plt.show()
```

```python
x = np.array([i for i in range(4)])

y1 = np.array([1,0,2,1])
y2 = np.array([0,2,1,0])

font = {'family': 'serif', 'color': 'black'}

plt.plot(x,y1,x,y2,marker='o')
plt.xlabel('Epochs', fontdict=font)
plt.ylabel('Score', fontdict=font)
plt.title('Performance(Accuracy)', fontdict=font, loc='left')
plt.grid()
plt.show()
```

```python
x = np.array([0,1,2,3])
y = np.array([3,8,1,10])

plt.subplot(1,2,1)
plt.plot(x,y)

x = np.array([0,1,2,3])
y = np.array([10,20,30,40])

plt.subplot(1,2,2)
plt.plot(x,y)

plt.show()
```

```python
x = np.linspace(0,5,4)
y = np.array([30,80,10,100])
y2 = np.array([80,30,90,50])

plt.subplot(2,1,1)
plt.plot(x,y,x,y2)
plt.title('Acc',loc='left')
plt.ylabel('Score')

# x = np.linspace(0,6,4)
y = np.linspace(60,100,4)
y2 = np.linspace(80,60,4)

plt.subplot(2,1,2)
plt.plot(x,y,x,y2)
plt.title('AUC',loc='left')
plt.xlabel('Epochs')
plt.ylabel('Score')

plt.show()
```

```python
x = np.array([i for i in range(6)])
y = np.array([0,8,12,20,26,38])

plt.scatter(x,y)
plt.show()
```

```python
x = np.array([i for i in range(6)])
y = np.array([0,2,8,1,14,7])
colors = ['red','green','purple','lime','aqua','yellow']
sizes = [10,60,120,80,20,190]

plt.scatter(x,y,color=colors,s=sizes,alpha=0.3)
plt.show()
```

```python
x = np.array([0,1,2,4,6])
yline1 = np.array([40,50,60,40,60])
yline2 = np.array([0,10,20,0,20])
ypoints = np.array([35,25,30,15,45])

plt.scatter(x,ypoints)
plt.plot(x,yline1,x,yline2)
plt.show()
```
